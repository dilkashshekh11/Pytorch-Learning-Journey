```python
# Create Tensor

import torch
x = torch.tensor([1,2,3,4,5]) # 1-D tensor tensor
print(x)
```

```python
y = torch.tensor([ [1,2],  # 2-D ensor t
                  [3,4]
                  ])
print(y)
```

```python
# zero tensor
y = torch.zeros(3,2) # rows x columns
print(y)
```

```python
# one tensor
o = torch.ones(3,2)  # rows x columns
print(o)
```

```python
# Random decimal tensor
r = torch.rand(3,2)  # rows x columns
print(r)
```

```python
# Shape - check rows x columns 
a = torch.tensor([ [1,2,3],
                  [4,5,6] 
                  ])
print(a.shape)
```

```python
# Indexing

 # 1- D tensor
# i = torch.tensor([10,20,30]) 
# print(i[0])
# print(i[1])
# print(i[2])
 
# 2- D tensor
d2 = torch.tensor([
    [10,20,30],
    [40,50,60]
])

print(d2[0,1])
print(d2[1,2])

```

```python
# Basic tensor operations
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
# Tensor and Numpy share memory 
import numpy as np 
arr = np.array([10,20,30])
t = torch.from_numpy(arr)

print(arr)
print(t)
```
