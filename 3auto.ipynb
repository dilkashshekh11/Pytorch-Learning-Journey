```python
# Computation Graph
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad= True)

y_pred = w * x
y_true = torch.tensor(10.0)

loss = (y_pred - y_true) **2 
print(loss)

loss.backward()
print(w.grad)

# print("Prediction : ",y_pred)  # with extra details

print("Prediction : ",y_pred.item())  # remove extra details
print("Loss : ", loss.item())
print("w.grad : ", w.grad.item())

# just understand 

# w.grade > 0 - decrease w
# w.grade < 0 - inclrease w

# just understand (gradient descent):
# new_value = old_value - (learn_Rate * gradient)
```

```python
import torch
# input fixed 
x = torch.tensor(2.0)

# learnable number 
w = torch.tensor(3.0, requires_grad= True)

# correct answer
y_true = torch.tensor(10.0)

# learning rate (10 %)
lr = 0.1  

y_pred = w * x

loss = (y_pred - y_true) ** 2

loss.backward()

print("Befor update")
print("w:", w.item())
print("Loss:", loss.item())
print("gradient:", w.grad.item())

with torch.no_grad():  # do not track 
      w -= lr * w.grad 

# Important : weight ( Reset weight)
w.grad.zero_()

print("\nAfter updated")
print("w:", w.item())
```

```python
x = torch.tensor([1., 2., 3., 4.])
w = torch.tensor(0.0, requires_grad= True)

y_true = torch.tensor([2., 4., 6., 8.])

lr = 0.1  #( 10 % learning rate)

epochs = 10

for epoch in range(epochs):
    # step 1 : prediction
    y_pred = w * x

    # step 2 : calculate loss
    loss = ((y_pred - y_true) ** 2).mean()
    loss.backward()

    #step 3 :

    with torch.no_grad():
        w -= lr * w.grad

    print(f"Epoch {epoch+ 1}: w={w.item():.4f},loss ={loss.item():.4f}")
```
