```python
import torch 

# step 1 input data x
x = torch.tensor([1., 2., 3., 4.])

# step 2 output data y 
y  = torch.tensor([2., 4., 6., 8.])

# step 3 wrong guesses for w and b 
w = torch.tensor(0.0)
b = torch.tensor(0.0)

# step 4 
y_pred = w * x + b

loss = ((y_pred - y) ** 2).mean()

print("Predicted Values: ", y_pred)
print("Actual Values : ", y)
print("Loss : ", loss)
```
